# Dataset
The dataset used in the repository is sourced from the following study:
B. Wicht and J. Hennebert, "Camera-based Sudoku recognition with deep belief network," 2014 6th International Conference of Soft Computing and Pattern Recognition (SoCPaR), Tunis, Tunisia, 2014, pp. 83-88, doi: 10.1109/SOCPAR.2014.7007986. keywords: {Training;Computer architecture;Image recognition;Transforms;Microprocessors;Image segmentation;Standards;Camera-based OCR;Deep Belief Network;Text Detection;Text Recognition},



```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mexwell/sudoku-image-dataset")

print("Path to dataset files:", path)
```

```python
import shutil
from pathlib import Path

data_dir = Path("./data")

shutil.copytree(src=path, dst=data_dir, dirs_exist_ok=True)
```
