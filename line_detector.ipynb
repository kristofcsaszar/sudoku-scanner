```python
from PIL import Image
from numpy import asarray

def load_image_as_np_array(path="data/v2_train/v2_train/image1.jpg"):
    img = Image.open(fp=path)
    return asarray(img)

img_array = load_image_as_np_array()
print(type(img_array))
print(img_array.shape)
```

```python
from matplotlib import pyplot as plt
import numpy as np

def display_image(img: np.ndarray):
    if img.ndim == 3:
        plt.imshow(img, interpolation='nearest')
    else:
        plt.imshow(img, cmap='gray')
    plt.show()
display_image(img=img_array)
```

```python
def grayscale(img: np.ndarray, weights: np.ndarray):
    weighed_img = img*weights
    h,w,ch = img.shape
    gr = np.zeros((h,w))
    for i in range(ch):
        gr+= weighed_img[:,:,i]

    return gr

uniform_w = [1/3, 1/3, 1/3]
human_w = [0.3, 0.59, 0.11]


display_image(grayscale(img_array, uniform_w))
display_image(grayscale(img_array, human_w))
```
