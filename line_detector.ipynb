```python
from PIL import Image
from numpy import asarray

def load_image_as_np_array(path="data/v2_train/v2_train/image1.jpg"):
    img = Image.open(fp=path)
    return asarray(img)

img_array = load_image_as_np_array()
print(type(img_array))
print(img_array.shape)
```

```python
from matplotlib import pyplot as plt
import numpy as np

def display_image(img: np.ndarray):
    if img.ndim == 3:
        plt.imshow(img, interpolation='nearest')
    else:
        plt.imshow(img, cmap='gray')
display_image(img=img_array)
```

```python
def grayscale(img: np.ndarray, weights: np.ndarray):
    r = img[:,:,0]
    g = img[:,:,1]
    b = img[:,:,2]

    r_adj=r*weights[0]
    g_adj = g*weights[1]
    b_adj =b*weights[2]

    return r_adj+g_adj+b_adj

w = [1/3, 1/3, 1/3]

gr_img = grayscale(img_array, w)
print(gr_img.shape)
print(gr_img.ndim)
display_image(gr_img)
```
